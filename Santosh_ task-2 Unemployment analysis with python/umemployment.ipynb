Heatmap: correlation between unemployment rate, employment rate, and labour participation rate

```python
correlation_matrix=df.select_dtypes('number').corr()
```

```python
sns.heatmap(data=correlation_matrix,annot=True,cmap='coolwarm',fmt='.2f')
```

### Observation

The correlation heatmap shows the relationships between unemployment rate, 
employment rate, and labour participation rate.

The correlation values range from -1 to +1. A value close to +1 indicates a 
strong positive relationship, while a value close to -1 indicates a strong 
negative relationship. A value close to 0 indicates a weak or no linear 
relationship.

The heatmap helps identify which variables are positively or negatively 
associated with unemployment. However, correlation does not imply causation.

Time-series line chart: unemployment rate over time for at least 3 major states/regions

```python
states=["Maharashtra","Karnataka","Tamil Nadu"]
```

```python
plt.figure(figsize=(14, 6))

for state in states:
    state_data = df[df['Region'] == state]
    
    plt.plot(
        state_data['Date'],
        state_data['Estimated Unemployment Rate (%)'],
        marker='o',
        label=state
    )

plt.title('Unemployment Rate Over Time')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.legend()
plt.grid()
```

### Observation

The time-series chart compares unemployment rates over time for three selected 
states or regions. The unemployment rates fluctuate throughout the study period, 
and each region follows a different pattern.

A noticeable change can be observed around the COVID-19 period, reflecting the 
impact of the pandemic on employment conditions. The chart also shows that the 
magnitude and timing of changes in unemployment varied between regions.

Bar chart: top 10 states with highest average unemployment rate  and EDA: Region-wise average unemployment rates

```python
average_employe_rate=df.groupby('Region')["Estimated Unemployment Rate (%)"].mean()
```

```python
average_employe_rate
```

### Observation

The region-wise analysis shows that the average unemployment rate varies across 
different states or regions. Some regions have higher average unemployment rates 
than others, indicating that unemployment was not equally distributed across 
the country during the study period.

The differences in average unemployment rates may reflect variations in 
economic activity, employment opportunities, and the impact of different 
economic conditions across regions.

```python
top_10_states=average_employe_rate.sort_values(ascending=False).head()
```

```python
top_10_states
```

```python
plt.figure(figsize=(12, 6))

plt.bar(top_10_states.index, top_10_states.values)

plt.title('Top 10 States with Highest Average Unemployment Rate')
plt.xlabel('State')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```

### Observation

The bar chart presents the 10 states with the highest average unemployment 
rates during the study period. The chart makes it easy to compare unemployment 
levels between these states.

The state at the top of the chart has the highest average unemployment rate 
among the selected top 10 states. The variation between the states indicates 
that unemployment levels differed considerably across regions.

Pre-COVID vs. post-COVID comparison (split data by date, calculate mean rates for each period)

```python
df['Date'] = pd.to_datetime(df['Date'])

pre_covid = df[df['Date'] < '2020-03-01']
post_covid = df[df['Date'] >= '2020-03-01']

pre_covid_mean = pre_covid['Estimated Unemployment Rate (%)'].mean()
post_covid_mean = post_covid['Estimated Unemployment Rate (%)'].mean()

print("Average Unemployment Rate - Pre-COVID:", round(pre_covid_mean, 2), "%")
print("Average Unemployment Rate - Post-COVID:", round(post_covid_mean, 2), "%")
```

### Observation

The comparison between the pre-COVID and post-COVID periods shows a difference 
in the average unemployment rate between the two periods.

The analysis helps us understand how unemployment conditions changed following 
the COVID-19 pandemic. Comparing the mean unemployment rates provides an 
overall view of the effect of the pandemic period on employment conditions.

EDA: Month-wise unemployment trends

```python
df['Date'] = pd.to_datetime(df['Date'])
df['Month'] = df['Date'].dt.month
monthly_avg = df.groupby('Month')['Estimated Unemployment Rate (%)'].mean()
print(monthly_avg)
```

### Observation

The month-wise analysis shows how the average unemployment rate changes across 
different months. The unemployment rate fluctuates throughout the year, with 
some months showing higher unemployment rates than others.

This variation suggests that unemployment may have seasonal or time-related 
patterns. Further analysis can help identify the months in which unemployment 
tends to be relatively high or low.
